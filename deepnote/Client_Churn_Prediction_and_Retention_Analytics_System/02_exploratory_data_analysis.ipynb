  # Client Churn Prediction and Retention Analytics System

## Phase 3 — Exploratory Data Analysis

### Objective

Understand the characteristics and behavioral patterns
associated with client churn.

### Key Questions

1. What is the overall churn rate?
2. Which client characteristics are associated with churn?
3. Which services are associated with retention?
4. Does contract type influence churn?
5. How do tenure and monthly charges relate to churn?
6. Which variables should receive attention during modeling?

```
import pandas as pd

DATA_PATH = "data/processed/cleaned_clients.csv"

df = pd.read_csv(DATA_PATH)

print("Dataset loaded:", df.shape)
```

```
# 1. Overall churn counts
print(df["Churn"].value_counts())

# 2. Churn percentages
print("\nChurn percentage:")
print(df["Churn"].value_counts(normalize=True) * 100)

# 3. Basic numerical summary
df[["tenure", "MonthlyCharges", "TotalCharges"]].describe()
```

## 1. Overall Churn Distribution

Before investigating individual features, we examine the
overall distribution of retained and churned clients.

```
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(7, 5))

sns.countplot(data=df, x="Churn")

plt.title("Client Churn Distribution")
plt.xlabel("Churn")
plt.ylabel("Number of Clients")

plt.show()
```

```
contract_churn = pd.crosstab(
    df["Contract"],
    df["Churn"],
    normalize="index"
) * 100

contract_churn
```

```
contract_churn.plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Churn Rate by Contract Type")
plt.xlabel("Contract Type")
plt.ylabel("Percentage (%)")
plt.xticks(rotation=0)

plt.show()
```

```
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="Churn",
    y="tenure"
)

plt.title("Tenure Distribution by Churn Status")
plt.xlabel("Churn")
plt.ylabel("Tenure (Months)")

plt.show()
```

```
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="Churn",
    y="MonthlyCharges"
)

plt.title("Monthly Charges by Churn Status")
plt.xlabel("Churn")
plt.ylabel("Monthly Charges")

plt.show()
```

```
df.groupby("Churn")[["tenure", "MonthlyCharges", "TotalCharges"]].mean()
```
