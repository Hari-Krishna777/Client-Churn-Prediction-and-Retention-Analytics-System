# Phase 2 — Data Cleaning Investigation

```
print("TotalCharges data type:", df["TotalCharges"].dtype)

print("\nBlank TotalCharges values:")
print((df["TotalCharges"].str.strip() == "").sum())
```

```
blank_total_charges = df[
    df["TotalCharges"].str.strip() == ""
]

blank_total_charges
```

```
blank_total_charges[
    ["customerID", "tenure", "MonthlyCharges", "TotalCharges", "Churn"]
]
```

```
df_clean = df.copy()

print("Original shape:", df.shape)
print("Working dataset shape:", df_clean.shape)
```

```
df_clean["TotalCharges"] = pd.to_numeric(
    df_clean["TotalCharges"].str.strip(),
    errors="coerce"
)
```

```
print(df_clean["TotalCharges"].dtype)
```

```
print("Missing TotalCharges after conversion:",
      df_clean["TotalCharges"].isna().sum())
```

```
df_clean["TotalCharges"] = df_clean["TotalCharges"].fillna(0)
```

```
print("Missing TotalCharges:",
      df_clean["TotalCharges"].isna().sum())

print("TotalCharges dtype:",
      df_clean["TotalCharges"].dtype)
```

```
df_clean[df_clean["tenure"] == 0][
    ["customerID", "tenure", "MonthlyCharges", "TotalCharges", "Churn"]
]
```

```
print(df_clean.columns.tolist())
```

```
print("=" * 50)
print("DATA CLEANING VALIDATION")
print("=" * 50)

print("Shape:", df_clean.shape)

print("\nMissing values:")
print(df_clean.isnull().sum())

print("\nDuplicate rows:")
print(df_clean.duplicated().sum())

print("\nData types:")
print(df_clean.dtypes)
```

```
import os

os.makedirs("data/processed", exist_ok=True)

processed_path = "data/processed/cleaned_clients.csv"

df_clean.to_csv(processed_path, index=False)

print(f"Saved cleaned dataset to: {processed_path}")
```
